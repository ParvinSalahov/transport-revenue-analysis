```python
transport_cols = [
    "Dəmir yolu (min manat)",
    "Dəniz (min manat)",
    "Hava (min manat)",
    "Avtomobil (min manat)",
    "Avtobus (min manat)",
    "Taksi (min manat)",
    "Metro (min manat)"
]

data_2019 = df[df["Year"] == 2019].iloc[0]
data_2020 = df[df["Year"] == 2020].iloc[0]

changes = {}

for col in transport_cols:
    pct_change = ((data_2020[col] - data_2019[col]) / data_2019[col]) * 100
    changes[col] = pct_change

changes_df = pd.DataFrame(
    changes.items(),
    columns=["Nəqliyyat növü", "Dəyişmə (%)"]
)

changes_df = changes_df.sort_values("Dəyişmə (%)")

print(changes_df)
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    changes_df["Nəqliyyat növü"],
    changes_df["Dəyişmə (%)"]
)

plt.axvline(0, color='black')
plt.xlabel("Dəyişmə (%)")
plt.title("2019–2020 Pandemiya Təsiri")

plt.show()
```

```python
transport_cols = [
    "Dəmir yolu (min manat)",
    "Dəniz (min manat)",
    "Hava (min manat)",
    "Avtomobil (min manat)",
    "Avtobus (min manat)",
    "Taksi (min manat)",
    "Metro (min manat)"
]

data_2020 = df[df["Year"] == 2020].iloc[0]
data_2024 = df[df["Year"] == 2024].iloc[0]

growth = {}

for col in transport_cols:
    growth[col] = ((data_2024[col] - data_2020[col]) / data_2020[col]) * 100

growth_df = pd.DataFrame(
    growth.items(),
    columns=["Nəqliyyat növü", "Artım (%)"]
)

growth_df = growth_df.sort_values("Artım (%)", ascending=False)

print(growth_df)
```

```python
plt.figure(figsize=(10,6))

bars = plt.barh(
    growth_df["Nəqliyyat növü"],
    growth_df["Artım (%)"]
)

plt.xlabel("Artım (%)")
plt.title("2020–2024 dövründə nəqliyyat sahələrinin bərpası")

for bar in bars:
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height()/2,
        f"{bar.get_width():.1f}%",
        va='center'
    )

plt.show()
```

```python
data_2019 = df[df["Year"] == 2019].iloc[0]

recovery = {}

for col in transport_cols:
    recovery[col] = ((data_2024[col] - data_2019[col]) / data_2019[col]) * 100

recovery_df = pd.DataFrame(
    recovery.items(),
    columns=["Nəqliyyat növü", "2019-a görə dəyişmə (%)"]
)

print(recovery_df.sort_values("2019-a görə dəyişmə (%)", ascending=False))
```
